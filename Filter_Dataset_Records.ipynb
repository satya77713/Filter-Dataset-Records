```python
setosa = df[df['Species'] == 'Iris-setosa']

display(setosa.head())

print("Number of records:", len(setosa))
```

```python
long_sepal = df[df['SepalLengthCm'] > 5]

display(long_sepal.head())

print("Number of records:", len(long_sepal))
```

```python
filtered_and = df[
    (df['PetalLengthCm'] > 4) &
    (df['PetalWidthCm'] > 1)
]

display(filtered_and.head())

print("Number of records:", len(filtered_and))
```

```python
filtered_or = df[
    (df['SepalLengthCm'] > 6) |
    (df['PetalLengthCm'] > 5)
]

display(filtered_or.head())

print("Number of records:", len(filtered_or))
```
